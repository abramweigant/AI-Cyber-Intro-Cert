## 1. Why machine learning in security, and where it is oversold

Security operations have a volume problem. A mid-sized network produces millions of log
lines a day, and somewhere in them are the few that matter. Hand-written rules — *block this
address, alert on that signature* — are precise and easy to audit, and they catch only what
somebody already thought to write a rule for. Analysts cannot read everything else.

Machine learning is used where a rule is hard to write but examples are easy to collect.
Nobody can write down the rule that separates a phishing URL from a legitimate one. It is
quite possible to collect eighty thousand of each and let a program find regularities.
That is what you will do in this course.

It is also oversold, in three recurring ways. Each is a module of this course.

- **The headline number means less than it appears to.** A detector that reports 99%
  accuracy may be measuring something other than what it is named for. You will produce
  several such numbers yourself and take them apart (Modules 2, 3 and 5).
- **The data is not what its documentation says.** The dataset in this module reports zero
  missing values and has missing values in 66 of its 111 columns.
- **The adversary adapts.** A spam filter is not judged against the weather. Someone is
  trying to get past it, and they can test their attempts. Module 7 puts you in that seat.

None of this makes the tools useless. It makes them tools: worth having, and only as
trustworthy as the checking done on them.

## 2. AI, machine learning, deep learning

The three terms are nested, and they are often used as if they were interchangeable.

| term | what it means | example in this course |
|---|---|---|
| **Artificial intelligence (AI)** | any system doing a task we would call intelligent if a person did it. The broadest term; includes hand-written rules. | a firewall rule set is AI in this old, wide sense |
| **Machine learning (ML)** | the part of AI where the behaviour is **learned from examples** instead of being written by hand | the phishing-URL classifier of Modules 2 and 3 |
| **Deep learning (DL)** | the part of ML that uses **neural networks with many layers** | the malware-image and email-text models of Module 5; the language models of Module 6 |

So every deep learning system is a machine learning system, and every machine learning
system is AI, and not the other way round.

**What "learning from examples" involves.** Whatever the method, the work has three stages,
and this course spends time on all three.

1. **Collect and prepare data.** Logs, URLs, emails, binaries, with or without labels. This
   stage decides more than the other two together, and it is where most of the trouble in
   this course will come from. This module is entirely about it.
2. **Fit a model.** An algorithm adjusts its internal numbers until its outputs match the
   examples. Different problems suit different algorithms; you will meet each one when you
   first use it.
3. **Evaluate.** Test the model on data it has not seen. The usual measures are accuracy,
   precision, recall and F1, which Module 2 introduces and Module 3 examines closely. A
   model that misses attacks (**false negatives**) and one that raises false alarms
   (**false positives**) fail in different ways, and which is worse depends on where it is
   deployed.

## 3. Three kinds of machine learning

They differ in what the examples come with.

| kind | the examples come with | it answers | a security example | where in this course |
|---|---|---|---|---|
| **Supervised** | the right answer for each one (a **label**) | "which known category is this?" | 88,647 URLs, each marked phishing or legitimate; learn to mark new ones | Modules 2, 3, 5 |
| **Unsupervised** | nothing | "what groups are in here, and what fits none of them?" | a month of network flows with no labels; find the hosts that behave unlike the rest | Module 4 |
| **Reinforcement** | a reward or penalty after each action | "what should I do next?" | an automated penetration-testing agent that tries actions against a simulated network and is rewarded for reaching a target host | not taught here; named so you recognise it |

Two refinements you will need.

**Supervised learning comes in two forms.** **Classification** predicts a category:
phishing or not, which malware family. **Regression** predicts a number: how many failed
logins to expect in the next hour. Every supervised model in this course is a classifier,
and most are **binary** classifiers, choosing between two classes.

**Unsupervised learning also comes in two forms.** **Clustering** groups similar rows.
**Dimensionality reduction** compresses many columns into a few so that the data can be
plotted and its structure seen. Module 4 uses both.

The practical question is always the first column of that table. If somebody has already
labelled examples of what you are looking for, you can use supervised learning, and it is
usually much the more accurate. If nobody has — a new kind of attack, a new network — you
cannot, however much you would like to.

## 4. Deep learning, briefly

A **neural network** is built from simple units, each of which takes some numbers in,
weights them, adds them up and passes the result on. Units are arranged in **layers**, and
each layer's outputs feed the next. "Deep" refers to having many layers.

What the layers buy is that the network can **work out its own features**. A classical model
needs a person to decide what to measure: the dataset in this module's lab is 111
measurements somebody chose to take of each URL. A deep network can be given something
rawer — the characters of a domain name, the bytes of a file — and learn which patterns in
it matter. Module 5 does exactly that with domain names, malware binaries and email text.

The cost is data and computation: these models have many more adjustable numbers and need
many more examples to set them. And they are no more immune than any other model to
learning the wrong thing from flawed data. In Module 5 a deep network will score 99.6% on
phishing emails, and you will work out what it actually learned.

On an ordinary table of numbers, deep learning is often *not* the best choice. You will
measure that too.

## 5. Binary classification, and its two ways of being wrong

Most detection problems in security are **binary classification**: phishing or legitimate,
malware or benign, intrusion or normal. A binary classifier can be wrong in two ways.

- A **false positive** flags something harmless. One costs an analyst a few minutes. A
  thousand a day teaches the whole team to ignore the alerts.
- A **false negative** misses the real thing.

There is a further difficulty that shapes everything from Module 3 onwards. Attacks are
**rare**. If one login in ten thousand is malicious, a "detector" that never raises an alert
is right 99.99% of the time and has detected nothing. So a single accuracy figure says very
little about a security model, and you will learn to ask for more than one.

## 6. Where machine learning is applied in security

Four areas, each with the kind of result you will read in papers and vendor material, and
the question this course will teach you to ask about it.

**Network security.** Firewalls, intrusion detection and access control protect traffic in
transit. Traditional systems match traffic against signatures of known attacks. ML-based
systems instead learn what traffic normally looks like and flag departures from it, or
learn from labelled attack traffic. Research prototypes report strong results; for example,
a retrainable firewall built on neural networks (Ahmadi, 2025, in the sources). *The
question to ask:* was the model tested on traffic from a different network and a different
month than it was trained on? Modules 4, 5 and 7 work with network flow data.

**Digital forensics.** After an incident, investigators reconstruct what happened from
logs, disk images and packet captures. It is slow, manual work, and ML is used to
prioritise what a person should look at. One study reports that a Random Forest trained on
packet captures identified attackers with near-perfect accuracy (Du, 2025).

> **Hold on to that phrase — "near-perfect accuracy."** You will meet claims like it
> repeatedly in security ML, including later in this module and again in Module 5.
> Sometimes they are real. Often they are an artifact of how the data was split, labelled,
> or collected. By the end of this course you will know which questions to ask before
> believing one. Start asking them now.

Module 6 ends with a forensic exercise: a week of logs, a language model to help triage
them, and a ground truth to check it against.

**Threat intelligence.** Teams collect reports, indicators and forum chatter to anticipate
attacks. The volume is far beyond what people can read, and language models are used to
extract and summarise. *The question to ask:* when the model summarises a report, is every
statement in the summary actually in the report? Module 6 tests that directly.

**Social engineering.** Phishing remains the most common way in. Classifiers are trained on
URLs (this module's dataset, which Modules 2 and 3 train on) and on message text (Modules 5 and 6). A recent
paper reports that a large language model distinguishes phishing from legitimate email with
over 97% accuracy (Lee, 2025). *The question to ask:* phishing emails and legitimate emails
from what sources, collected when? Module 5 shows a case where the answer changes
everything.

### 4.1 Find the impossible number

Read the `min` row, and then read the `mean` row for `qty_params`.

`qty_params` counts how many query parameters a URL has. A count. You cannot have minus one
of something — and yet the minimum is `-1`, and **the average is negative**.

An average count below zero is not a surprising result. It is an impossible one, and
impossible results mean your data does not contain what you think it contains.

`-1` instead of being left blank.

`isnull()` cannot see this. To pandas, `-1` is just a number, and it will happily average
it in with the real ones.

## Module 1 Summary: what the audit found

You loaded one dataset and ran perhaps a dozen commands. Here is what they turned up:

| check | result |
|---|---|
| rows and columns | 88,647 URLs, 111 features + 1 label |
| `isnull()` | **0 missing values** |
| actual missing values | **66 of 111 columns**, encoded as `-1` |
| worst affected | the 19 `_params` columns — missing in **91.6%** of rows |
| effect on a summary statistic | mean `params_length` reads **5.27** instead of **73.93** |
| exact duplicate rows | 1,438 |
| feature vectors carrying **both** labels | 6 (13 rows) |
| columns with no information at all | 13, every one of them a `_domain` count |
| class balance | 34.6% phishing |

Nothing here means the dataset is unusable. Most of it is perfectly workable once you know.
The point is that **you had to look.** The command everyone runs — `isnull()` — reported a
spotlessly clean dataset, and it was wrong in the most consequential way available: it was
reassuring.

That pattern is the spine of this course. In Module 3 you will find that the strongest
signal in this very dataset is an accident of how it was collected. In Module 5 you will
meet a preprocessing step that quietly copies rows across a train/test split, a benchmark
containing two classes that are literally the same image, and a phishing detector that
scores 99.6% while solving a different problem than the one it is named for.

The skill you started building today is not pandas syntax. It is the habit of asking what a
number would look like if it were wrong.
